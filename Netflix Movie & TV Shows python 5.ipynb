```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="type")

plt.title("Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
```

```python
type_percentage = df["type"].value_counts(normalize=True) * 100

type_percentage
```

```python
df["rating"].value_counts()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    y="rating",
    order=df["rating"].value_counts().index
)

plt.title("Netflix Content by Rating")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")

plt.show()
```

```python
df["release_year"].describe()
```

```python
year_count = df["release_year"].value_counts().sort_index()

year_count.tail(20)
```

```python
plt.figure(figsize=(14,6))

year_count.plot()

plt.title("Netflix Content Release Trend")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
recent_years = df[df["release_year"] >= 2010]

recent_years["release_year"].value_counts().sort_index()
```

```python
plt.figure(figsize=(12,5))

sns.countplot(
    data=recent_years,
    x="release_year"
)

plt.title("Netflix Content Released Since 2010")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10,7))

df["country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Countries by Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```
